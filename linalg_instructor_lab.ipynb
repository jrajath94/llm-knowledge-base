## Section 1 — Vectors, Dot Product, Norms

### Task 1.1: Implement dot + L2 norm (no np.linalg.norm)

- dot(x,y) = sum(x_i * y_i)
- ||x||_2 = sqrt(dot(x,x))

**Explain:** What does dot product measure geometrically?

**Hints**
- dot(x,y) = sum(x_i * y_i)
||x||_2 = sqrt(dot(x,x))

**Explain:** What does dot product measure geometrically?

```python
def dot(x, y):
    return float(np.sum(x * y))

def l2_norm(x):
    return float(np.sqrt(dot(x, x)))

x = np.array([1.,2.,3.]); y=np.array([4.,5.,6.])
check('dot', abs(dot(x,y) - 32.0) < 1e-9)
check('norm', abs(l2_norm(x) - np.sqrt(14.0)) < 1e-9)

# Rationale: dot measures alignment (||x|| ||y|| cosÎ¸). Norm is vector magnitude.
```

```python
# Checks
check('dot', abs(dot(x,y) - 32.0) < 1e-9)
check('norm', abs(l2_norm(x) - np.sqrt(14.0)) < 1e-9)
```

**Why this works.** dot measures alignment (||x|| ||y|| cosÎ¸). Norm is vector magnitude.

## Task 4: Project vector v onto vector u
*Section: Projections (Least Squares Intuition)*

## Section 3 — Projections (Least Squares Intuition)

### Task 3.1: Project vector v onto vector u

proj_u(v) = (u^T v / u^T u) * u

- Use your dot()

**Explain:** Why does projection show up in linear regression?

**Hints**
- Use your dot()

**Explain:** Why does projection show up in linear regression?

```python
def proj(u, v):
    denom = dot(u, u)
    if denom == 0:
        raise ValueError('u must be non-zero')
    return (dot(u, v) / denom) * u

u=np.array([1.,0.,0.]); v=np.array([2.,3.,4.])
check('proj', np.allclose(proj(u,v), np.array([2.,0.,0.])))
# Rationale: projection is the building block of least squares (closest point in a subspace).
```

```python
# Checks
check('proj', np.allclose(proj(u,v), np.array([2.,0.,0.])))
```

**Why this works.** projection is the building block of least squares (closest point in a subspace).

## Task 5: Solve min_w ||Xw - y||^2

## Section 4 — Least Squares (Closed Form)

### Task 4.1: Solve min_w ||Xw - y||^2

Use normal equation: w = (X^T X)^{-1} X^T y

- Use `np.linalg.solve` (more stable than explicit inverse)

**FAANG gotcha:** Don’t compute matrix inverse unless you must.

![LU Decomposition Overview](/images/lu-decomposition-overview.png)

**Hints**
- Use `np.linalg.solve` (more stable than explicit inverse)

```python
n,d=50,3
X=rng.standard_normal((n,d))
w_true=np.array([1.5,-2.0,0.7])
y=X@w_true + 0.01*rng.standard_normal(n)

XtX = X.T @ X
Xty = X.T @ y
w_hat = np.linalg.solve(XtX, Xty)
err = np.linalg.norm(w_hat - w_true)
print('err', err)
check('close', err < 0.2)

# Rationale: solve is more stable than explicit inverse; inverse is slower + numerically worse.
```

```python
# Checks
check('close', err < 0.2)
```

**Why this works.** solve is more stable than explicit inverse; inverse is slower + numerically worse.

## Section 5 — Eigenvalues & SVD Intuition

### Task 5.1: PCA via SVD

Steps:
1. Center X
2. Compute SVD: X = U S V^T
3. Take top-k components from V

- `U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)`

**Explain:** Why does SVD show up in embeddings and dimensionality reduction?

![Eigen Decomposition Overview](/images/eigen-decomposition-overview.png)

![Singular Value Decomposition (SVD)](/images/svd-overview.png)

**Hints**
- `U, S, Vt = np.linalg.svd(X_centered, full_matrices=False)`

**Explain:** Why does SVD show up in embeddings and dimensionality reduction?

```python
X = rng.standard_normal((200,10))
Xc = X - X.mean(axis=0, keepdims=True)
U,S,Vt = np.linalg.svd(Xc, full_matrices=False)
k=3
W = Vt.T[:, :k]
Z = Xc @ W
X_recon = Z @ W.T
recon_err = np.linalg.norm(Xc - X_recon) / np.linalg.norm(Xc)
print('relative recon error', recon_err)
check('shapes', W.shape == (10,k) and Z.shape == (200,k))

# Rationale: SVD gives orthonormal directions capturing maximal variance (PCA).
# In ML, low-rank structure appears in embeddings, recommender matrices, and compression.
```

```python
# Checks
check('shapes', W.shape == (10,k) and Z.shape == (200,k))
```

**Why this works.** SVD gives orthonormal directions capturing maximal variance (PCA).
