## Section 1 — Convolution shape math

Output height/width for conv2d: 
- `H_out = floor((H + 2P - kH)/S) + 1`
- `W_out = floor((W + 2P - kW)/S) + 1`

### Task 1.1
Implement `conv2d_out_shape`.

Hint: use integer division `//` for floor behavior.

```python
def conv2d_out_shape(H, W, kH, kW, stride=1, padding=0):
    H_out = (H + 2*padding - kH) // stride + 1
    W_out = (W + 2*padding - kW) // stride + 1
    return int(H_out), int(W_out)

check('shape_case1', conv2d_out_shape(8, 8, 3, 3, stride=1, padding=0) == (6, 6))
check('shape_case2', conv2d_out_shape(8, 8, 3, 3, stride=1, padding=1) == (8, 8))
check('shape_case3', conv2d_out_shape(8, 8, 5, 5, stride=2, padding=0) == (2, 2))
```

## Section 3 — MaxPool forward

### Task 3.1
Implement 2×2 maxpool with stride 2 for NCHW tensors.

```python
def maxpool2d_forward(x, pool=2, stride=2):
    N, C, H, W = x.shape
    H_out = (H - pool) // stride + 1
    W_out = (W - pool) // stride + 1
    out = np.zeros((N, C, H_out, W_out), dtype=float)
    for n in range(N):
        for c in range(C):
            for i in range(H_out):
                hs = i * stride
                for j in range(W_out):
                    ws = j * stride
                    patch = x[n, c, hs:hs+pool, ws:ws+pool]
                    out[n, c, i, j] = float(np.max(patch))
    return out

z = rng.normal(size=(2, 3, 8, 8))
p = maxpool2d_forward(z, pool=2, stride=2)
check('pool_shape', p.shape == (2, 3, 4, 4))
```

```python
# Checks
check('pool_shape', p.shape == (2, 3, 4, 4))
```

**Why this works.** ## Section 3 — MaxPool forward

Rationale: pooling trades resolution for invariance (and loses some information).
