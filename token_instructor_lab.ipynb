## Section 0 - Corpus + normalization

```python
CORPUS = [
  'The quick brown fox jumps over the lazy dog.',
  'Tokenization matters: e-mail, emails, emailing; 1,000 and 3.14 and 2e10.',
  'New York-based startups re-entered the market in 2024-2025.',
  "We're testing edge-cases: unicode-ish dashes - and quotes 'like this'." ,
  'Subword tokenizers handle rarewordsbetter than word tokenizers.',
]

def normalize_text(s: str) -> str:
    return re.sub(r'\s+', ' ', s).strip()

_TOK_RE = re.compile(r'(?:[A-Za-z]+(?:\'[A-Za-z]+)?)|(?:\d+(?:,\d{3})*(?:\.\d+)?(?:e[+-]?\d+)?)|(?:[^\s])')

def pretokenize(s: str):
    s = normalize_text(s)
    return _TOK_RE.findall(s.lower())

def is_wordlike(t: str) -> bool:
    return bool(re.match(r'^[a-z0-9]', t)) and any(ch.isalnum() for ch in t)

WORDS = []
for s in CORPUS:
    for t in pretokenize(s):
        if is_wordlike(t):
            WORDS.append(t)
print('n_words', len(WORDS), 'unique', len(set(WORDS)))
check('words_nonempty', len(WORDS) > 0)
```
