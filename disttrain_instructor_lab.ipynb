## Task 2: 0 — Synthetic regression dataset

## 0 — Synthetic regression dataset

Create a simple regression problem so we can study optimization and timing without framework complexity.

```python
def make_regression(n: int = 20_000, d: int = 50, noise: float = 0.2, seed: int = 0):
    rng = np.random.default_rng(seed)
    X = rng.normal(size=(n, d)).astype(np.float64)
    w_true = rng.normal(size=(d,)).astype(np.float64)
    y = (X @ w_true + noise * rng.normal(size=n)).astype(np.float64)
    return X, y, w_true

X, y, w_true = make_regression(seed=0)
print('X', X.shape, 'y', y.shape)
check('shapes', X.shape[0] == y.shape[0] and X.shape[1] == w_true.shape[0])
```

**Why this works.** A convex regression problem is enough to show the distributed-training timing tradeoffs without needing a deep network.
