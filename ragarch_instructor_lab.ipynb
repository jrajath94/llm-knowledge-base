## Section 3 — Retrieval (relevance + optional freshness)

```python
def embed_query(q: str):
    qc = vectorize_counts([q], w2i)
    qv = apply_tfidf(qc, idf)
    return l2_normalize_rows(qv)[0]

def retrieve(q: str, k: int = 3, freshness_lambda: float = 0.0):
    qv = embed_query(q)
    sims = E @ qv
    ts = np.array([c["ts"] for c in chunks], dtype=np.float64)
    tsn = (ts - ts.min()) / (ts.max() - ts.min() + 1e-12)
    score = sims + freshness_lambda * tsn
    idx = np.argsort(-score)[:k]
    return idx.tolist(), score[idx].tolist()

print('demo:', retrieve('production access mfa policy', k=3, freshness_lambda=0.0)[0])
```

## Task 5: Section 4 — Context assembly under a budget (keep citations)

## Section 4 — Context assembly under a budget (keep citations)

```python
def pack_context(indices, max_chars: int = 240, include_citations: bool = True):
    parts = []
    used = 0
    for ix in indices:
        c = chunks[ix]
        header = f"[{c['chunk_id']}|ts={c['ts']}] " if include_citations else ""
        piece = header + c["text"]
        if used + len(piece) + 1 > max_chars:
            break
        parts.append(piece)
        used += len(piece) + 1
    return "\n".join(parts)

ctx = pack_context(retrieve("mfa production access", k=4)[0], max_chars=240)
print(ctx)
check('ctx_nonempty', len(ctx) > 0)
```
